```python
import time
import json
from retrieve import *

sparse = create_bm25_retriever()
vector_store = create_vector_store()

with open("new_test_set2.json", 'r') as file:
    test_set = json.load(file)
test_set = test_set['semantics']

precision = 0
recall=0
for item in test_set:
    gold_ids = item['new_chunks']
    docs = retrieve_hybrid_chunks(item['question'], sparse, vector_store)
    ids_from_docs = [doc.metadata['chunk_id'] for doc in docs]
    print("guessed: ",ids_from_docs)
    print("actual: ",gold_ids)

    hits = len(set(ids_from_docs) & set(gold_ids))
    print(f"Correct {hits}/{len(gold_ids)}")
    print("Hits: ", hits)
    recall += hits / len(gold_ids)
    print("recall: ", hits/len(gold_ids))
    print('\n')
    time.sleep(1)

recall = recall/len(test_set)
print("Recall: ", recall)
```
